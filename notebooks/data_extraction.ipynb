```python
import pandas as pd
from datetime import date

# Load the Kaggle LinkedIn postings dataset
df = pd.read_csv("D:/projects/fake jd/data/raw/archive/postings.csv")  # adjust path/filename as needed

# Filter to job categories comparable to your fraud sample (WFH / data entry / entry-level roles)
keywords = ["data entry", "remote", "work from home", "virtual assistant", 
            "form filling", "online job", "part time", "assistant"]
mask = df['title'].str.contains('|'.join(keywords), case=False, na=False)
filtered = df[mask].dropna(subset=['title'])

# Random sample - adjust n to how many you want
sample = filtered.sample(n=min(80, len(filtered)), random_state=42).reset_index(drop=True)

# Map into your exact column schema
out = pd.DataFrame({
    "ID": range(1, len(sample) + 1),
    "Platform": "LinkedIn (Kaggle dataset)",
    "Account Handle": sample['company_name'].fillna("N/A"),
    "Job Title": sample['title'],
    "Company (as advertised)": sample['company_name'].fillna("N/A"),
    "Work Mode": sample['formatted_work_type'].fillna("N/A"),
    "Shifts Listed": "N/A",
    "Salary Range (INR/month)": sample.apply(
        lambda r: f"{r['min_salary']} - {r['max_salary']}" 
        if pd.notna(r['min_salary']) and pd.notna(r['max_salary']) else "Not specified",
        axis=1
    ),
    "Requirements Listed": sample['description'].str.slice(0, 200).fillna("N/A"),  # truncated snippet
    "Target Audience": "General applicants",
    "Contact Channel": "LinkedIn application portal",
    "Red Flags Identified": "None",
    "Red Flag Count": 0,
    "Brand Impersonation?": "No",
    "Label": "Real",
    "Notes": "Sourced from Kaggle LinkedIn Job Postings dataset (arshkon, 2023-2024)",
    "Date Collected": str(date.today()),
    "Source File": "kaggle_linkedin_job_postings.csv"
})

out.to_csv("legit_postings_mapped.csv", index=False)
print(f"Saved {len(out)} rows to legit_postings_mapped.csv")
```
